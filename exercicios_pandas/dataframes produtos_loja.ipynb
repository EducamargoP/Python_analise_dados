```python
# criação de DataFrame através de um dicionairio
```

```python
dados = {'Produtos':['Videogame', 'PC', 'Teclado', 'Mouse', 'Microfone'], 'PREÇO':[2600.00, 2450.90, 99.90, 75, 129.99],'Custo':[2000,1950.99,54,39.80,79.75]}
```

```python
df3
```
