# MERGE / JOIN / CONCATENATE

```python
# CONCATENATE -> melhor utilizado quando temos colunas com nomes iguais, melhor utilizado na vertical
# MERGE / JOIN -> horizontal ultilizando o HOW passando onde unir (left, inner, outer, right)
# outer -> junta todos os que nao tem valor ele deixa NULO, inner junta apenas que tem em ambas tabelas
```

```python
import pandas as pd
import numpy as np
```

```python
df1 = pd.DataFrame({'A':['A0','A1','A2','A3'],'B':['B0','B1','B2','B3'],'C':['C0','C1','C2','C3'],'D':['D0','D1','D2','D3']},index=[0,1,2,3])
```

```python
df2 = pd.DataFrame({'A':['A4','A5','A6','A7'],'B':['B4','B5','B6','B7'],'C':['C4','C5','C6','C7'],'D':['D4','D5','D6','D7']}, index=[4,5,6,7])
```

```python
df3 = pd.DataFrame({'A':['A8','A9','A10','A11'],'B':['B8','B9','B10','B11'],'C':['C8','C9','C10','C11'],'D':['D8','D9','D10','D11']},index=[8,9,10,11])
```

```python
dicionario = {'ID':[10,11,12,13,14],'NOME':['Maria','Joana','Claudia','Lucas','Gabriel'],'CIDADE':['SP','RJ','SP','CP','SC']}
```

```python
dicionario2 = {'ID':[16,11,12,13,18],'EXPERIENCIA':['junior','senior','pleno','estagiario','analista']}
```

```python
df4 = pd.DataFrame(data=dicionario)
df5 = pd.DataFrame(data=dicionario2)
```

```python
#cocatenar
```

```python
df1
```

```python
df2
```

```python
df3
```

```python
df_concat = pd.concat([df1,df2,df3])
```

```python
df_concat
```

```python
# MERGE
```

```python
df4
```

```python
df5
```

```python
#merge // inner lembrando que so aparece id iguais
```

```python
merge_inner = pd.merge(df4,df5)
```

```python
merge_inner
```

```python
# merge outer // lembrando que esse aparece todos , os que nao tem numero aparece nulo
```

```python
merge_outer = pd.merge(df4,df5, how='outer')
```

```python
merge_outer
```

```python
# merge left // esse os valores da esquerda que manda
```

```python
merge_left = pd.merge(df4,df5,how='left')
```

```python
merge_left
```

```python
# merge right // todos os valores da direita que manda 
```

```python
merge_right = pd.merge(df4,df5,how='right')
```

```python
merge_right
```

```python
# join 
```

```python
join_esq = df4.join(df5, lsuffix='_x', rsuffix='Y')
```

```python
join_esq
```
