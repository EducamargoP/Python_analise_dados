```python
for index, row in df.iterrows():
    time_casa = row["TimeCasa"]
    time_fora = row["TimeFora"]
    gols_casa = row["GolCasa"]
    gols_fora = row["GolFora"]
   
    
    #VITORIAS DOS mandantes
    if gols_casa > gols_fora:
        df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'PONTOS'] += 3 
        df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'VITÓRIAS'] += 1
        df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'DERROTAS'] += 1
        
    #vitoria do vicitante    
    elif gols_fora > gols_casa:
        df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'PONTOS'] += 3 
        df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'VITÓRIAS'] += 1
        df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'DERROTAS'] += 1 
        
    #empates
    else:
        df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'PONTOS'] += 1
        df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'PONTOS'] += 1
        df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'EMPATE'] += 1
        df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'EMPATE'] += 1
    
    # atualização saldo de gols
    df_classificacao.loc[df_classificacao['TIME'] == time_casa, 'SALDO G'] += (gols_casa - gols_fora)
    df_classificacao.loc[df_classificacao['TIME'] == time_fora, 'SALDO G'] += (gols_fora - gols_casa)
    
  
    
```

```python
#ordernar por pontos e saldo de gols
df_classificacao.sort_values(by=['PONTOS','SALDO G'], ascending=False, inplace=True)

#resetar o indice para ficar mais limpo
df_classificacao.reset_index(drop=True, inplace=True)
```
