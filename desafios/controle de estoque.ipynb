🎯 Novo desafio: Controle de estoque
Imagine que você tem uma loja com 12 produtos. Cada produto tem:

Estoque inicial

Entradas (quantidade que chegou)

Saídas (quantidade vendida)

Você precisa calcular:

Saldo final = Estoque inicial + Entradas – Saídas

Situação:

Se saldo final = 0 → “Esgotado”

Se saldo final < 5 → “Baixo estoque”

Caso contrário → “Disponível”
