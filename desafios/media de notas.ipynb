🎯 Novo desafio: Sistema de notas escolares
Imagine que você tem uma turma de alunos e precisa montar uma tabela de desempenho, como se fosse uma “classificação” igual ao campeonato.

O que você deve implementar:
Média de notas → cada aluno tem notas em várias provas.

Situação final → aprovado se média ≥ 7, recuperação se média entre 5 e 6.9, reprovado se < 5.

Ranking de desempenho → ordenar os alunos pela média, como se fosse uma tabela de classificação.

Coluna de faltas → se quiser complicar, adicione faltas e defina reprovação por excesso de faltas.

```python
import pandas as pd
import numpy as np
```

```python
alunos = {'alunos':['FELIPE', 'AMANDA', 'LEONARNDO', 'EDUARDO', 'MIGUEL', 'MARIA EDUARDA', 'LARISSA', 'HUGO', 'EDMAR', 'SANDRA', 'CONCEIÇÃO', 'MARIA CLARA'],'TRABALHO':[7,10,6,8,5,4,3,2,7,4,2,6],'TAREFAS':[10,9,7,6,6,5,4,3,5,4,2,3],'PROVA':[9,8,9,7,8,6,5,4,3,2,3,1]}
```

```python
sala_aula = pd.DataFrame(alunos)
```

```python
sala_aula
```

```python
sala_aula['menor nota'] = sala_aula[['TRABALHO', 'TAREFAS','PROVA']].min(axis=1)
```

```python
sala_aula['SITUAÇÃO'] = sala_aula['menor nota'].apply( lambda x: 'REPROVADO' if x < 6 else ('RECUPERAÇÃO' if x < 7 else 'APROVADO'))
```
