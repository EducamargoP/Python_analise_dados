```python
cont = 1 
```

```python
while (cont <= 10):
    print(cont)
    cont += 1
```

```python
# utilizando break
cont = 1 
```

```python
while (cont <= 10):
    print(cont)
    if (cont == 7):
        break
    cont += 1
```

```python
# utilizando o continue / nesse caso ele pula o valor adicionado na condição
```

```python
cont = 0
while (cont < 10):
    cont += 1
    if (cont == 7):
        continue
    print(cont)
```
